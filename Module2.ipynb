**MODULE 2:  Object-Oriented Programming**

**Class:**

A class is a blueprint for creating objects. It defines data and behavior.

Eg. A `Student` class can describe what every student object should contain.

```python
class Student:
    print("this")

student1 = Student()
print(student1)
print(type(student1))
```

**Object:**

An object is an instance created from a class.

If `Student` is the blueprint, `student1` is an actual student object.

```python
# 1. Define the class
class Dog:
    # The initializer method (sets up the properties for each new dog)
    def __init__(self, name, breed):
        self.name = name    # Property unique to each dog
        self.breed = breed  # Property unique to each dog

    # A method (a function that belongs to the dog)
    def bark(self):
        return f"{self.name} says Woof!"

# 2. Create actual objects (Instances) from the class
dog1 = Dog("Buddy", "Golden Retriever")
dog2 = Dog("Bella", "Poodle")

# 3. Access the properties and methods
print(dog1.name)       # Output: Buddy
print(dog2.breed)      # Output: Poodle
print(dog1.bark())     # Output: Buddy says Woof!
```

**Constructor – __init__()**

`__init__()` runs automatically when an object is created and initializes its data.

A student object can receive a name during creation.

```python
class Student:
    def __init__(self, name):
        self.name = name

student = Student("Anu")
print(student.name)
```

**Instance Variables:**

Instance variables belong to a particular object, so different objects can have different values.

Two students can have different names

```python
class Student:
    def __init__(self, name):
        self.name = name

s1 = Student("Anu")
s2 = Student("Ravi")

print(s1.name)
print(s2.name)
```

**Class Variables:**

A class variable is shared by instances unless an instance overrides it.

All students can share the same school name.

```python
class Student:
    school = "ABC College"

    def __init__(self, name):
        self.name = name

s1 = Student("Anu")
s2 = Student("Ravi")

print(s1.school)
print(s2.school)
```

**Instance Methods:**

An instance method operates on a particular object and normally receives `self`.

A student object can display its own name.

```python
class Student:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Student:", self.name)

o_student = Student("Anu")
o_student.display()
```

**Static Methods:**

A static method belongs logically to a class but does not need the object or class as its first argument.

A utility function that checks whether a number is even can be placed inside a related class.


```python
class NumberUtils:
    @staticmethod
    def is_even(number):
        return number % 2 == 0

print(NumberUtils.is_even(10))
```

**Class Methods:**

A class method receives the class as `cls` and can work with class-level data.

A class method can create an object from an alternative input format.

```python
class Student:
    school = "ABC College"

    def __init__(self, name):
        self.name = name

    @classmethod
    def school_name(cls):
        return cls.school

print(Student.school_name())
```

**Properties – @property:**

A property allows a method to be accessed like an attribute and is useful for controlled access or validation.

An employee's salary can be validated before it is stored

```python
class Employee:
    def __init__(self, salary):
        self._salary = salary

    @property
    def salary(self):
        return self._salary

    @salary.setter
    def salary(self, value):
        if value < 0:
            raise ValueError("Salary cannot be negative")
        self._salary = value

employee = Employee(30000)
employee.salary=-2
print(employee.salary)
```

**Encapsulation:**

Encapsulation means keeping data and related methods together and controlling how internal data is accessed.

An account balance should not be changed arbitrarily; deposit and withdrawal methods can control it.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance

account = BankAccount(1000)
account.deposit(500)
print(account.get_balance())
```

**Abstraction:**

Abstraction hides implementation details and exposes only the required interface.

A user calls `start()` without needing to know the internal engine-starting process.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        print("Car started")

car = Car()
car.start()
```

**Inheritance:**

Inheritance allows one class to reuse or extend another class.

A `Dog` can inherit common behavior from `Animal`.

```python
class Animal:
    def eat(self):
        print("Eating")

class Dog(Animal):
    def bark(self):
        print("Barking")

dog = Dog()
dog.eat()
dog.bark()
```

**Multiple Inheritance:**

A class can inherit from more than one parent class.

A smart device may have both camera and GPS capabilities.

```python
class Camera:
    def take_photo(self):
        print("Photo taken")

class GPS:
    def locate(self):
        print("Location found")

class SmartPhone(Camera, GPS):
    pass

phone = SmartPhone()
phone.take_photo()
phone.locate()
```

**Multilevel Inheritance:**

Multilevel inheritance forms a chain such as Grandparent → Parent → Child.

A `Dog` inherits from `Animal`, while `Puppy` inherits from `Dog`.

```python
class Animal:
    def eat(self):
        print("Eating")

class Dog(Animal):
    def bark(self):
        print("Barking")

class Puppy(Dog):
    def play(self):
        print("Playing")

p = Puppy()
p.eat()
p.bark()
p.play()
```

**Hierarchical Inheritance:**

Multiple child classes inherit from the same parent.

Both `Dog` and `Cat` inherit from `Animal`.

```python
class Animal:
    def eat(self):
        print("Eating")

class Dog(Animal):
    def bark(self):
        print("Barking")

class Cat(Animal):
    def meow(self):
        print("Meowing")

Dog().eat()
Cat().eat()
```

**Polymorphism:**

Polymorphism means the same interface can produce different behavior depending on the object.

Different animals can respond differently to the same `sound()` call.

```python
class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")

for animal in [Dog(), Cat()]:
    animal.sound()
```

**Method Overriding:**

A child class overrides a parent method by defining a method with the same name.

A generic animal sound can be replaced by a dog-specific sound.

```python
class Animal:
    def sound(self):
        print("Some sound")

class Dog(Animal):
    def sound(self):
        print("Bark")

Dog().sound()
```

**Method Overloading – Python Approach:**

Python does not support traditional compile-time method overloading with multiple definitions of the same method. A common Python approach is to use default arguments or `*args`.

One function can accept one or two numbers

```python
class Calculator:
    def add(self, a, b=0):
        return a + b

calc = Calculator()
print(calc.add(5))
print(calc.add(5, 10))
```

**Composition:**

Composition represents a strong "has-a" relationship where one object contains another object as a part of its functionality.

A car has an engine.

```python
class Engine:
    def start(self):
        print("Engine started")

class Car:
    def __init__(self):
        self.engine = Engine()

    def start(self):
        self.engine.start()

Car().start()
```

**Aggregation:**

Aggregation is a weaker "has-a" relationship where contained objects can exist independently.

A department has teachers, but teachers can exist independently of the department.

```python
class Teacher:
    def __init__(self, name):
        self.name = name

class Department:
    def __init__(self, teacher):
        self.teacher = teacher

teacher = Teacher("Anu")
department = Department(teacher)

print(department.teacher.name)
```

**Association:**

Association means two independent objects are connected because they interact.

A doctor and patient can interact without either being a component of the other.

```python
class Doctor:
    def consult(self, patient):
        print("Consulting", patient.name)

class Patient:
    def __init__(self, name):
        self.name = name

patient = Patient("Ravi")
Doctor().consult(patient)
```

**Magic / Dunder Methods:**

Dunder methods have names surrounded by double underscores and allow objects to work with Python's built-in operations.

`__len__()` controls what `len(object)` does.

```python
class Team:
    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)

team = Team(["Anu", "Ravi", "Meena"])
print(len(team))
```

**super():**

`super()` gives access to methods or initialization logic from a parent class.

A child class can reuse the parent's constructor.

```python
class Person:
    def __init__(self, name):
        self.name = name

class Employee(Person):
    def __init__(self, name, employee_id):
        super().__init__(name)
        self.employee_id = employee_id

employee = Employee("Anu", 101)
print(employee.name, employee.employee_id)
```

 **__str__():**

`__str__()` defines a user-friendly string representation of an object.

Printing an object can display meaningful information instead of a default memory-style representation.


```python
class Student:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Student: {self.name}"

print(Student("Anu"))
```

**__repr__():**

`__repr__()` should provide a useful developer-oriented representation of an object.

It can show the object's important fields.

```python
class Student:
    def __init__(self, name):
        self.name = name

    def __repr__(self):
        return f"Student(name={self.name!r})"

print(repr(Student("Anu")))
```

**__len__():**

`__len__()` defines the result of `len(object)`.

A custom collection can report its number of elements.

```python
class Course:
    def __init__(self, topics):
        self.topics = topics

    def __len__(self):
        return len(self.topics)

course = Course(["Python", "SQL", "ML"])
print(len(course))
```

**Dataclasses – Bonus:**

A dataclass reduces boilerplate code for classes that mainly store data.

Instead of manually writing an initializer and representation, `@dataclass` can generate them

```python
from dataclasses import dataclass

@dataclass
class Employee:
    name: str
    salary: float

employee = Employee("Anu", 35000)
print(employee)
```

**Singleton Pattern:**

A Singleton pattern is designed so that a class has only one shared instance.

A configuration object can be shared across an application.

```python
class Singleton:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

a = Singleton()
b = Singleton()

print(a is b)
```
