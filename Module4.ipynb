**Module 4: Logging:**

**What is Logging?**

Logging records application events, errors, warnings, and useful diagnostic information in a structured way.

Instead of relying only on screen output, a production application can save errors to a log file.


```python
import logging

logging.basicConfig(level=logging.INFO)
logging.info("Application started")
```

**Why use Logging?**

Logging helps developers monitor applications, diagnose failures, understand program behavior, and maintain production systems.

A failed database connection can be recorded with an error message.


```python
import logging

logging.basicConfig(level=logging.INFO)

try:
    10 / 0
except ZeroDivisionError:
    logging.error("Division by zero occurred")
```

**print() vs logging:**

`print()` is simple and mainly intended for direct output. Logging supports levels, timestamps, handlers, files, formatting, and configurable production behavior.

Use `print()` for a quick temporary message and logging for application events

**Log Levels:**

Log levels communicate the importance of an event:
- `DEBUG` – detailed diagnostic information
- `INFO` – normal application events
- `WARNING` – unexpected but non-fatal situation
- `ERROR` – an operation failed
- `CRITICAL` – serious failure that may stop the application


```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.debug("Debug details")
logging.info("Application started")
logging.warning("Low disk space")
logging.error("Database operation failed")
logging.critical("Application cannot continue")
```

**Log Formatters:**

A formatter controls how log records appear.

We can include time, level, and message.

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logging.info("Formatted log message")
```

**File Handlers:**

A file handler writes log messages to a file.

Production errors can be saved for later investigation.

```python
import logging

logger = logging.getLogger("app")
logger.setLevel(logging.INFO)

handler = logging.FileHandler("application.log")
logger.addHandler(handler)

logger.info("Saved to file")
```

**Console Handlers:**

A console handler sends logs to the terminal.

Developers can see application events while the program is running.

```python
import logging

logger = logging.getLogger("console_app")
logger.setLevel(logging.INFO)

handler = logging.StreamHandler()
logger.addHandler(handler)

logger.info("Visible in console")
```

**Rotating Logs:**

Rotating logs prevent one log file from growing indefinitely by creating backup files when the size limit is reached.

A server can keep several small log files instead of one huge file.

```python
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger("rotating")
logger.setLevel(logging.INFO)

handler = RotatingFileHandler(
    "app.log",
    maxBytes=10000,
    backupCount=3
)

logger.addHandler(handler)
logger.info("Rotating log example")
```

**Logging Best Practices:**

Good logging should be meaningful, consistent, and safe.

Useful production logs include timestamps and levels, while passwords and secrets should never be logged.

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(name)s | %(message)s"
)

logger = logging.getLogger(__name__)
logger.info("Model training started")
```
