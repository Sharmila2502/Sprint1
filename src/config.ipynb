```python
class ConfigManager:
    """Manage application configuration."""

    def __init__(self):
        self.config = {
            "database": "stackly_db",
            "host": "localhost",
            "username": "root",
            "output_folder": "outputs"
        }

    def get(self, key):
        """Get a configuration value."""

        return self.config.get(key)
```

```python
config = ConfigManager()
```

```python
print(config.get("database"))
print(config.get("host"))
print(config.get("username"))
print(config.get("output_folder"))
```
