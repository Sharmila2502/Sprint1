```python
import csv
import json
from pathlib import Path
```

```python
class InvalidFileException(Exception):
    """Raised when a file is invalid or unsupported."""
    
    pass
```

```python
class DataLoader:
    """Load data from CSV, JSON, and TXT files."""

    def validate_file(self, file_path: str) -> Path:
        """Validate the file path and return a Path object."""

        path = Path(file_path)

        if not path.exists():
            raise InvalidFileException(
                f"File does not exist: {file_path}"
            )

        if not path.is_file():
            raise InvalidFileException(
                f"Path is not a file: {file_path}"
            )

        return path

    def load_csv(self, file_path: str) -> list[dict]:
        """Load and return data from a CSV file."""

        path = self.validate_file(file_path)

        if path.suffix.lower() != ".csv":
            raise InvalidFileException(
                "Expected a CSV file."
            )

        with open(path, "r", encoding="utf-8") as file:
            reader = csv.DictReader(file)
            return list(reader)

    def load_json(self, file_path: str):
        """Load and return data from a JSON file."""

        path = self.validate_file(file_path)

        if path.suffix.lower() != ".json":
            raise InvalidFileException(
                "Expected a JSON file."
            )

        with open(path, "r", encoding="utf-8") as file:
            return json.load(file)

    def load_txt(self, file_path: str) -> str:
        """Load and return the contents of a TXT file."""

        path = self.validate_file(file_path)

        if path.suffix.lower() != ".txt":
            raise InvalidFileException(
                "Expected a TXT file."
            )

        with open(path, "r", encoding="utf-8") as file:
            return file.read()

    def load(self, file_path: str):
        """Load data automatically based on the file extension."""

        path = self.validate_file(file_path)

        extension = path.suffix.lower()

        if extension == ".csv":
            return self.load_csv(file_path)

        if extension == ".json":
            return self.load_json(file_path)

        if extension == ".txt":
            return self.load_txt(file_path)

        raise InvalidFileException(
            f"Unsupported file format: {extension}"
        )
```

```python
loader = DataLoader()
```

```python
try:
    data = loader.load("data/unknown.csv")
    print("File loaded successfully")

except InvalidFileException as error:
    print("File error:", error)

finally:
    print("Loading operation completed")
```
