```python
class ConfigManager:
    """Manage application configuration settings."""

    def __init__(self, config: dict):
        self._config = config

    @property
    def database_host(self) -> str:
        """Return the database host."""

        return self._config["database_host"]

    @property
    def database_user(self) -> str:
        """Return the database username."""

        return self._config["database_user"]

    @property
    def database_name(self) -> str:
        """Return the database name."""

        return self._config["database_name"]

    @property
    def log_file(self) -> str:
        """Return the log file path."""

        return self._config["log_file"]

    @property
    def output_folder(self) -> str:
        """Return the output folder path."""

        return self._config["output_folder"]

    def get(self, key: str, default=None):
        """Return a configuration value by key."""

        return self._config.get(key, default)

    @classmethod
    def create_default(cls):
        """Create ConfigManager with default settings."""

        default_config = {
            "database_host": "localhost",
            "database_user": "root",
            "database_name": "stackly_db",
            "log_file": "logs/application.log",
            "output_folder": "outputs"
        }

        return cls(default_config)
```

```python
config = ConfigManager.create_default()
```
