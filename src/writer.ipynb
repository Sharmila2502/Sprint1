```python
import csv
import json
import pickle
from pathlib import Path
```

```python
class DataWriter:
    

    def write_csv(self, data, file_path):
        """Write a list of dictionaries to a CSV file."""

        Path(file_path).parent.mkdir(
            parents=True,
            exist_ok=True
        )

        with open(
            file_path,
            "w",
            newline="",
            encoding="utf-8"
        ) as file:

            csv_writer = csv.DictWriter(
                file,
                fieldnames=data[0].keys()
            )

            csv_writer.writeheader()
            csv_writer.writerows(data)

    def write_json(self, data, file_path):
        """Write data to a JSON file."""

        Path(file_path).parent.mkdir(
            parents=True,
            exist_ok=True
        )

        with open(
            file_path,
            "w",
            encoding="utf-8"
        ) as file:

            json.dump(
                data,
                file,
                indent=4
            )

    def write_pickle(self, data, file_path: str):
        """Write a Python object to a Pickle file."""

        Path(file_path).parent.mkdir(
            parents=True,
            exist_ok=True
        )

        with open(
            file_path,
            "wb"
        ) as file:

            pickle.dump(
                data,
                file
            )

    def write_txt(self, data: str, file_path: str):
        """Write text data to a TXT file."""

        Path(file_path).parent.mkdir(
            parents=True,
            exist_ok=True
        )

        with open(
            file_path,
            "w",
            encoding="utf-8"
        ) as file:

            file.write(data)
```

```python
writer = DataWriter()
```

```python
customer_data = [
    {
        "id": 1,
        "name": "Arun",
        "city": "Coimbatore",
        "balance": 50000
    },
    {
        "id": 2,
        "name": "Meena",
        "city": "Ooty",
        "balance": 75000
    },
    {
        "id": 3,
        "name": "Ravi",
        "city": "Chennai",
        "balance": 40000
    }
]
```

TEST CSV

```python
writer.write_csv(
    customer_data,
    "outputs/customers_output.csv"
)
```

```python
print(
    Path("outputs/customers_output.csv").read_text(
        encoding="utf-8"
    )
)
```

TEST JSON:

```python
writer.write_json(
    customer_data,
    "outputs/customers_output.json"
)
```

```python
print(Path("outputs/customers_output.json").read_text(encoding="utf-8"))
```

TEST PICKLE

```python
writer.write_pickle(
    customer_data,
    "outputs/customers_output.pkl"
)
```

```python
print(Path("outputs/customers_output.pkl").exists())
```

```python
notes = """This is a sample text file.
DataWriter is being tested.
The file is created using Python.
"""
```

```python
writer.write_txt(
    notes,
    "outputs/notes_output.txt"
)
```
