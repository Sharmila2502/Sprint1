```python
from pathlib import Path

Path("logs").mkdir(
    parents=True,
    exist_ok=True
)
```

1.Get the logger called application
2.Set the logging level
3.Create a file handler [where the log messages should go]
4.Set the handler's level
5.Create the log format



```python


class LoggerManager:
    """Manage application logging."""

    @staticmethod
    def setup_logger():
        """Configure and return the application logger."""

        log_path = Path("logs/application.log")

        log_path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        logger = logging.getLogger("application")
        logger.setLevel(logging.INFO)

        file_handler = logging.FileHandler(
            log_path,
            encoding="utf-8"
        )

        file_handler.setLevel(logging.INFO)

        formatter = logging.Formatter(
            "%(asctime)s - %(levelname)s - %(message)s"
        )

        file_handler.setFormatter(formatter)

        logger.addHandler(file_handler)

        return logger
```

```python
logger = LoggerManager.setup_logger()
```

```python
logger.info("Application started successfully")
logger.warning("This is a warning message")
logger.error("An error occurred while processing the file")
```
