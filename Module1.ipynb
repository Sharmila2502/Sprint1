BASIC CONCEPTS:

Variables:

A variable is a name used to refer to a value stored by a Python program. Python does not require us to declare the data type separately

Eg.A variable is like a labeled box. If the label is `age`, the box can contain the value `25`


```python
name = "ABC"
age = 25

print(name)
print(age)
```

Datatypes:

A datatype tells Python what kind of value an object represents. Common types are `int`, `float`, `str`, `bool`, `list`, `tuple`, `set`, and `dict`.

A whole number such as `10` is an integer, while `10.5` is a floating-point number.

```python
age = 25
salary = 35000.50
name = "Anu"
is_active = True

#`type()` shows the type of each value
print(type(age))
print(type(salary))
print(type(name))
print(type(is_active))
```

Type Casting:

Type casting means converting a value from one data type to another.

Input from `input()` is text. If we want to calculate with it, we can convert it to `int` or `float`.


```python

#`int()` and `float()` convert strings into numeric values
age_text = "25"
age = int(age_text)

price_text = "99.50"
price = float(price_text)

print(age + 5)
print(price + 10)
```

CONTROL FLOW:

if / elif / else:

Conditional statements allow a program to choose different actions based on conditions.

A student's result can be classified as pass, distinction, or fail depending on marks.

```python
marks = 75

if marks >= 80:
    print("Distinction")
elif marks >= 40:
    print("Pass")
else:
    print("Fail")
```

for loop:

A `for` loop repeats code for each item in an iterable.

If we have five student names, we can process each name one by one

```python
students = ["Anu", "Ravi", "Meena"]

for student in students:
    print(student)
```

while loop:

A `while` loop repeats while a condition remains true.

Eg.A counter can continue until it reaches 5

```python
count = 1
#The condition is checked before every iteration
while count <= 5:
    print(count)
    count += 1
```

break:

`break` immediately stops the current loop.

Stop searching once a target value is found.


```python
numbers = [3, 7, 10, 15, 20]
#The loop terminates when `number` becomes 10
for number in numbers:
    if number == 10:
        break
    print(number)
```

continue:

`continue` skips the current iteration and moves to the next one.

We can skip even numbers and print only odd numbers.


```python
#Even numbers are skipped
for number in range(1, 6):
    if number % 2 == 0:
        continue
    print(number)
```

List:

A list is an ordered, mutable collection. Duplicate values are allowed.

A shopping cart can change as products are added or removed.

```python
cart = ["Pen", "Book", "Bag"]
cart.append("Bottle")
cart.remove("Pen")

print(cart)
```

Tuple:

A tuple is an ordered collection that cannot normally be modified after creation.

Eg. Coordinates `(10, 20)` can be represented as a tuple.

Set:

A set stores unique values and is useful for set operations.

If a list contains duplicate customer IDs, a set can remove duplicates.

```python
customer_ids = [101, 102, 101, 103, 102]
unique_ids = set(customer_ids)
#Duplicate IDs disappear when converted to a set
print(unique_ids)
```

Dictionary:

A dictionary stores data as key-value pairs.

A customer record can map `"name"` to a name and `"age"` to an age.


```python
customer = {
    "name": "Anu",
    "age": 25,
    "city": "Coimbatore"
}
#Keys are used to access values
print(customer["name"])
customer["age"] = 26
print(customer)
type(customer)
```

List comprehension is a compact way to create a list from an iterable.

Eg. Create squares from 1 to 5.

```python
#The expression is evaluated for every number in the range
squares = [number ** 2 for number in range(1, 6)]
print(squares)
```

Dictionary Comprehension:

Dictionary comprehension creates dictionaries using a compact expression.

Create a dictionary mapping numbers to their squares.

```python
#Each number becomes a key and its square becomes the value
squares = {number: number ** 2 for number in range(1, 6)}
print(squares)
```

FUNCTIONS:

Function Definition:

A function is a reusable block of code designed to perform a particular task.

Instead of writing greeting code repeatedly, define it once and call it whenever needed.


```python
#`def` defines the function and `greet()` calls it
def greet():
    print("Welcome to Python")

greet()
```

Arguments:

Arguments are values supplied to a function when it is called.

A function can receive two numbers and calculate their sum.

```python
def add(a, b):
    return a + b

result = add(10, 20)
print(result)
```

Default Arguments:

A default argument has a value that is used when the caller does not provide one.

A greeting can default to `"Guest"`

```python

def greet(name="Guest"):
    print("Hello", name)

greet()     #The first call uses the default value
greet("Anu")
```

Keyword Arguments:

Keyword arguments pass values by parameter name.

The order of keyword arguments does not have to match the function definition.

```python
def student(name, age):
    print(name, age)

student(age=22, name="Ravi")   #`name=` and `age=` explicitly identify the parameters
```

Variable-Length Arguments – *args :

`*args` allows a function to accept any number of positional arguments.

A function can calculate the sum of any number of values.


```python
def total(*numbers):      #`numbers` becomes a tuple containing all positional arguments.
    return sum(numbers)

print(total(10, 20))
print(total(10, 20, 30, 40))
```

Variable-Length Arguments – **kwargs

`**kwargs` allows a function to accept any number of keyword arguments.

A profile function can accept different named details.


```python
def show_profile(**details):
    for key, value in details.items():
        print(key, ":", value)

show_profile(name="Anu", city="Coimbatore", role="Data Scientist")
```

Lambda Functions:

A lambda is a small anonymous function, generally used for short operations.

A lambda can square a number without defining a full function

```python
square = lambda x: x ** 2     # `x ** 2` is returned automatically
print(square(5))
```

Recursive Functions:

A recursive function calls itself to solve a smaller version of the same problem. It must have a base condition.

Factorial of 5 can be represented as `5 * 4 * 3 * 2 * 1`

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

ADVANCED PYTHON:

Iterators:

An iterator is an object that returns values one at a time using `__next__()`.

Instead of loading and processing every value at once, an iterator can provide one value at a time.


```python
numbers = iter([10, 20, 30])  #`iter()` creates an iterator and `next()` retrieves the next value

print(next(numbers))
print(next(numbers))
print(next(numbers))
```

Generators:

A generator produces values lazily instead of storing all generated values in memory at once.

A large sequence can be generated one number at a time.

```python
def numbers():
    for number in range(1, 4):   #The generator pauses after each `yield`
        yield number

for value in numbers():
    print(value)
```

yield:

`yield` returns a value from a generator while preserving its execution state.

After yielding 1, execution can resume later from the next line.


```python
def count_up_to(limit):
    number = 1
    while number <= limit:
        yield number          #Each `yield` pauses the function
        number += 1

for value in count_up_to(3):
    print(value)
```

Decorators:

A decorator is a function that adds or changes behavior of another function without changing its original code.

A decorator can print a message before and after a function executes.

```python
def log_call(func):
    def wrapper():
        print("Function started")
        func()
        print("Function finished")
    return wrapper

@log_call                           #`@log_call` wraps `greet()` with additional behavior
def greet():
    print("Hello")

greet()
```

Closures:

A closure is an inner function that remembers values from its enclosing function even after the outer function has finished.

A multiplier function can remember the multiplier.

```python
def multiplier(factor):
    def multiply(number):           #`multiply()` remembers `factor=2`.
        return number * factor
    return multiply

double = multiplier(2)
print(double(10))
```

global and nonlocal:

`global` allows a function to modify a module-level variable. `nonlocal` allows an inner function to modify a variable from its enclosing function.

Use `global` for a module-level counter and `nonlocal` for a nested counter.

```python
count = 0

def increase_global():
    global count       # `global` targets the module variable
    count += 1

increase_global()
print(count)


def counter():
    value = 0

    def increase():
        nonlocal value  #`nonlocal` targets the enclosing function variable 
        value += 1
        return value

    return increase

c = counter()
print(c())
print(c())
```

Type Hinting:

Type hints document the expected types of variables, parameters, and return values. Python does not normally enforce them at runtime.

A function can clearly indicate that it expects two integers and returns an integer.


```python
def add(a: int, b: int) -> int:     #`: int` describes expected input types and `-> int` describes the expected return type.
    return a + b

print(add(10, 20))
```

BUILT-IN FUNCTIONS:

map():

`map()` applies a function to every item in an iterable.

Eg. Converts a list of numbers into their squares.

```python
numbers = [1, 2, 3, 4]
squares = list(map(lambda x: x ** 2, numbers))
print(squares)
```

filter():

`filter()` keeps elements for which a condition is true.

Eg. Keeps only even numbers

```python
numbers = [1, 2, 3, 4, 5, 6]
even = list(filter(lambda x: x % 2 == 0, numbers))
print(even)
```

zip():

`zip()` combines elements from multiple iterables position by position.

Eg. Combine student names with marks.


```python
names = ["Anu", "Ravi", "Meena"]
marks = [80, 75, 90]

for name, mark in zip(names, marks):
    print(name, mark)
```

enumerate():

`enumerate()` provides both an index and a value while iterating.

Eg. To display student numbers along with names.

```python
students = ["Anu", "Ravi", "Meena"]

for index, student in enumerate(students, start=1):
    print(index, student)
```

sorted():

`sorted()` returns a new sorted list and does not modify the original iterable.

Eg. Sort marks from highest to lowest.

```python
marks = [70, 90, 60, 80]
result = sorted(marks, reverse=True)

print(result)
print(marks)
```

any() :

`any()` returns `True` if at least one item is truthy.

Eg. Check whether at least one mark is above 90.

```python
marks = [55, 60, 95, 70]
print(any(mark > 90 for mark in marks))
```

all() :

`all()` returns `True` only when every item is truthy.

Eg. Check whether every mark is at least 40.

```python
marks = [55, 60, 95, 70]
print(all(mark >= 40 for mark in marks))
```

isinstance() :

`isinstance()` checks whether an object belongs to a specified type or class.

Eg. Validate whether an input value is an integer.

```python
value = 25
print(isinstance(value, int))
print(isinstance(value, str))
```

id() :

`id()` returns an identity value associated with an object during its lifetime.

Eg. Two variables referring to the same list have the same object identity.

```python
numbers = [1, 2, 3]
same_numbers = numbers
num=['abc','bhd']

print(id(numbers))
print(id(num))
print(id(same_numbers))
print(id(numbers) == id(same_numbers))
```
