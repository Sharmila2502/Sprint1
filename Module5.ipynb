**Module 5 – File Handling**

**Reading Text Files:**
    
Reading a file loads its stored text so a program can process it.

A program can read a configuration or text file.

```python
with open("sample.txt", "r", encoding="utf-8") as file:
    content = file.read()

print(content)
```

**Writing Text Files:**

Writing stores text in a file. Mode `"w"` creates or replaces the file.

Save a program result into a report file.

```python
with open("result.txt", "w", encoding="utf-8") as file:
    file.write("Model accuracy: 82%")
```

**CSV Files:**

CSV stores tabular data as rows and columns separated by delimiters.

A dataset containing names and marks can be stored in CSV.


```python
import csv

rows = [
    ["Name", "Mark"],
    ["Anu", 80],
    ["Ravi", 75]
]

with open("students.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerows(rows)

with open("students.csv", "r", encoding="utf-8") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

**JSON Files**

JSON stores structured data using objects, arrays, and primitive values. It is commonly used for APIs and configuration.

A customer record can be stored as a JSON object.

```python
import json

customer = {
    "name": "Anu",
    "age": 25,
    "active": True
}

with open("customer.json", "w", encoding="utf-8") as file:
    json.dump(customer, file, indent=4)

with open("customer.json", "r", encoding="utf-8") as file:
    data = json.load(file)

print(data)
```

**Pickle Files:**

Pickle serializes Python objects into a byte stream so they can be saved and loaded later. **Never unpickle untrusted files**, because loading a malicious pickle can execute code.

A trained Python object can be serialized for later use.

```python
import pickle

model_data = {"model": "DecisionTree", "accuracy": 0.92}

with open("model.pkl", "wb") as file:
    pickle.dump(model_data, file)

with open("model.pkl", "rb") as file:
    loaded = pickle.load(file)

print(loaded)
```

**Context Manager – with:**

A context manager automatically handles setup and cleanup of a resource.

A file opened with `with` is automatically closed

```python
with open("sample.txt", "w", encoding="utf-8") as file:
    file.write("Hello")
```

**pathlib:**

`pathlib` provides an object-oriented way to work with file and directory paths.

Check whether a file exists.

```python
from pathlib import Path

path = Path("sample.txt")
print(path.exists())
print(path.suffix)
```

**os:**

The `os` module provides operating-system-related functions such as environment variables and directory operations.

Get the current working directory

```python
import os

print(os.getcwd())
print(os.listdir("."))
```

**shutil:**

`shutil` provides high-level operations for copying, moving, and removing files/directories.

Copy a file to a backup location

```python
import shutil

shutil.copy("sample.txt", "sample_backup.txt")
```

**File Validation:**

File validation checks whether a file exists and whether it meets requirements such as extension or size.

Before reading a CSV, verify that it exists.

```python
from pathlib import Path

file_path = Path("students.csv")

if not file_path.exists():
    print("File does not exist")
elif file_path.suffix.lower() != ".csv":
    print("Expected a CSV file")
else:
    print("File is valid")
```

**Directory Operations:**

Programs often need to create, inspect, rename, or remove directories.

Create a directory for model outputs.

```python
from pathlib import Path

output_dir = Path("outputs")
output_dir.mkdir(exist_ok=True)

print(output_dir.exists())
```
