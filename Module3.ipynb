**Module 3 – Exception Handling:**

**try:**

`try` contains code that may produce an exception.

A division operation may fail if the denominator is zero

```python
try:
    result = 10 / 2
    print(result)
except ZeroDivisionError:
    print("Cannot divide by zero")
```

**except:**

`except` catches and handles a specific exception.

Catch invalid integer conversion.

```python
try:
    number = int("abc")
except ValueError:
    print("Invalid number")
```

**else:**

`else` executes only when the `try` block completes without an exception.

Print the result only when division succeeds.

```python
try:
    result = 10 / 2
except ZeroDivisionError:
    print("Invalid division")
else:
    print("Result:", result)
```

**finally:**

`finally` runs whether an exception occurs or not. It is useful for cleanup.

A file or database connection can be closed in `finally`.

```python
try:
    print(10 / 2)
except ZeroDivisionError:
    print("Error")
finally:
    print("Cleanup completed")
```

**raise:**

`raise` allows the programmer to deliberately generate an exception when a business rule is violated.

Reject a negative age.

```python
age = -5

if age < 0:
    raise ValueError("Age cannot be negative")
```

**assert:**

`assert` checks an assumption. If the condition is false, Python raises `AssertionError`. It is mainly useful for debugging and internal assumptions, not for user input validation.

A machine-learning dataset should have at least one row.

```python
data = [10, 20, 30]

assert len(data) > 0, "Dataset cannot be empty"
print("Dataset is valid")
```

**Custom Exceptions:**

A custom exception is a programmer-defined exception representing an application-specific error.

A banking application can have an `InsufficientBalanceError`.

```python
class InsufficientBalanceError(Exception):
    pass

balance = 1000
withdraw = 1500

try:
    if withdraw > balance:
        raise InsufficientBalanceError("Insufficient balance")
except InsufficientBalanceError as error:
    print(error)
```

**Built-in Exceptions:**

**ValueError:**
**Occurs when:** a function receives a value of the correct general type but an invalid value

```python
try:
    number = int("hello")
except ValueError:
    print("ValueError handled")
```

**TypeError:**
**Occurs when:** an operation is performed on an inappropriate type.

```python
try:
    result = "10" + 5
except TypeError:
    print("TypeError handled")
```

**IndexError:**
**Occurs when:** a sequence index is outside its valid range

```python
try:
    numbers = [10, 20]
    print(numbers[5])
except IndexError:
    print("IndexError handled")
```

**KeyError:**
**Occurs when:** a dictionary key does not exist.

```python
try:
    student = {"name": "Anu"}
    print(student["age"])
except KeyError:
    print("KeyError handled")
```

**AttributeError:**
**Occurs when:** an object does not have the requested attribute or method.

```python
try:
    text = "hello"
    text.append("x")
except AttributeError:
    print("AttributeError handled")
```

**NameError:**
**Occurs when:** Python cannot find a variable or name

```python
try:
    print(total_amount)
except NameError:
    print("NameError handled")
```

**FileNotFoundError:**
**Occurs when:** a requested file does not exist.

```python
try:
    with open("missing.txt", "r") as file:
        print(file.read())
except FileNotFoundError:
    print("File not found")
```

**ZeroDivisionError:**
**Occurs when:** a number is divided by zero.


```python
try:
    print(10 / 0)
except ZeroDivisionError:
    print("Cannot divide by zero")
```

 **ImportError:**
**Occurs when:** an import statement cannot import the requested name from a module

```python
try:
    from math import unknown_function
except ImportError:
    print("ImportError handled")
```

**ModuleNotFoundError:**
**Occurs when:** Python cannot find the requested module.

```python
try:
    import module_that_does_not_exist
except ModuleNotFoundError:
    print("ModuleNotFoundError handled")
```
