```python
cursor.execute("""
CREATE TABLE departments (
    department_id INT PRIMARY KEY,
    department_name VARCHAR(50) NOT NULL UNIQUE,
    location VARCHAR(50))""")

connection.commit()

print("Departments table created!")
```

```python
cursor.execute("""
CREATE TABLE employees (
    employee_id INT PRIMARY KEY,
    employee_name VARCHAR(100) NOT NULL,
    department_id INT,
    salary DECIMAL(10,2),
    city VARCHAR(50),
    hire_date DATE,

    FOREIGN KEY (department_id)
    REFERENCES departments(department_id))""")

connection.commit()

print("Employees table created!")
```

```python
cursor.execute("""
CREATE TABLE projects (
    project_id INT PRIMARY KEY,
    project_name VARCHAR(100) NOT NULL,
    department_id INT,

    FOREIGN KEY (department_id)
    REFERENCES departments(department_id))""")

connection.commit()

print("Projects table created!")
```

**ORDER BY**

```python
cursor.execute("""SELECT employee_name, salary FROM employees ORDER BY salary DESC """)

rows = cursor.fetchall()

for row in rows:
    print(row)
```

**GROUP BY**

```python
cursor.execute("""SELECT department_id, COUNT(*) AS employee_count FROM employees GROUP BY department_id""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

**GROUP BY WITH SUM**

```python
cursor.execute("""SELECT department_id, SUM(salary) AS total_salary FROM employees GROUP BY department_id""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

**HAVING**

```python
cursor.execute("""SELECT department_id, COUNT(*) AS employee_count FROM employees GROUP BY department_id HAVING COUNT(*) > 1""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

**DISTINCT**

```python
cursor.execute("""SELECT DISTINCT city FROM employees""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

**LIMIT**

```python
cursor.execute("""SELECT * FROM employees LIMIT 5""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

**JOINS**

**INNER JOIN**

```python
cursor.execute("""
SELECT e.employee_name,
       e.salary,
       d.department_name
FROM employees e
INNER JOIN departments d
ON e.department_id = d.department_id
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

**LEFT JOIN**

```python
cursor.execute("""
SELECT
    e.employee_name,
    e.salary,
    d.department_name
FROM employees e
LEFT JOIN departments d
ON e.department_id = d.department_id
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

**RIGHT JOIN**

```python
cursor.execute("""
SELECT
    e.employee_name,
    d.department_name
FROM employees e
RIGHT JOIN departments d
ON e.department_id = d.department_id
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

**FULL JOIN**

```python
cursor.execute("""
SELECT
    e.employee_name,
    d.department_name
FROM employees e
LEFT JOIN departments d
ON e.department_id = d.department_id

UNION

SELECT
    e.employee_name,
    d.department_name
FROM employees e
RIGHT JOIN departments d
ON e.department_id = d.department_id
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

**CROSS JOIN**

```python
cursor.execute("""
SELECT
    e.employee_name,
    p.project_name
FROM employees e
CROSS JOIN projects p
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

**ADVANCED SQL**

**UNION**

```python
cursor.execute("""
SELECT city
FROM employees
WHERE department_id = 1

UNION

SELECT city
FROM employees
WHERE department_id = 5
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

**CASE**

```python
cursor.execute("""
SELECT
    employee_name,
    salary,
    CASE
        WHEN salary >= 80000 THEN 'High Salary'
        WHEN salary >= 60000 THEN 'Medium Salary'
        ELSE 'Low Salary'
    END AS salary_category
FROM employees
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

**SUBQUERY**

```python
cursor.execute("""
SELECT
    employee_name,
    salary
FROM employees
WHERE salary > (
    SELECT AVG(salary)
    FROM employees
)
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

**CTE**

```python
cursor.execute("""
WITH high_salary_employees AS (
    SELECT employee_name, salary
    FROM employees
    WHERE salary > 70000
)
SELECT *
FROM high_salary_employees
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

**VIEW**

```python
cursor.execute("""
CREATE OR REPLACE VIEW employee_details AS
SELECT
    e.employee_id,
    e.employee_name,
    e.salary,
    d.department_name,
    d.location
FROM employees e
JOIN departments d
ON e.department_id = d.department_id
""")

connection.commit()

print("View created!")
```

**INDEX**

```python
cursor.execute("""
CREATE INDEX idx_employee_salary
ON employees(salary)
""")

connection.commit()

print("Index created!")
```

**WINDOW FUNCTIONS**

**ROW_NUMBER()**

```python
cursor.execute("""
SELECT
    employee_name,
    salary,
    ROW_NUMBER() OVER (
        ORDER BY salary DESC
    ) AS row_num
FROM employees
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

**RANK**

```python
cursor.execute("""
SELECT
    employee_name,
    salary,
    RANK() OVER (
        ORDER BY salary DESC
    ) AS salary_rank
FROM employees
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

**DENSE_RANK()**

```python
cursor.execute("""
SELECT
    employee_name,
    salary,
    DENSE_RANK() OVER (
        ORDER BY salary DESC
    ) AS salary_rank
FROM employees
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

**LAG()**

```python
cursor.execute("""
SELECT
    employee_name,
    salary,
    LAG(salary) OVER (
        ORDER BY salary DESC
    ) AS previous_salary
FROM employees
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

**LEAD()**

```python
cursor.execute("""
SELECT
    employee_name,
    salary,
    LEAD(salary) OVER (
        ORDER BY salary DESC
    ) AS next_salary
FROM employees
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```
